# Create and populate the directory data

```python
import os
import pathlib
import subprocess
import sys
```

```python
os.chdir('../../..')
dirname='data'
if not os.path.isdir(dirname):
    subprocess.call(['mkdir',dirname])
os.chdir(dirname)
```

```python
indir_nutrients='/Volumes/max07/scripts/gud_groups/gud_1d_35+16/input_noradtrans/input'
nut_l=list(range(0,21,2))
nut_s_l=[str(nut).rjust(2,'0') for nut in nut_l]
```

## DataS1_winter_nutrients_in_situ_qik

```python
dirname='DataS1_winter_nutrients_in_situ_qik'
files=['1D_BB_NO3_GE_spring.delZ1016mm.32bits.bin',
       '1D_BB_Si_GE_spring.delZ1016mm.32bits.bin']
```

```python
if not os.path.isdir(dirname):
    subprocess.call(['mkdir',dirname])
os.chdir(dirname)
```

```python
for basename in files:
    absname=os.path.join(indir_nutrients,basename)
    subprocess.call(['ln','-fs',absname,basename])
os.chdir('..')
```

## DataS2_winter_nutrients_simulated

```python
dirname='DataS2_winter_nutrients_simulated'
nutrients=['NO3','Si']
```

```python
if not os.path.isdir(dirname):
    subprocess.call(['mkdir',dirname])
os.chdir(dirname)
```

```python
for nutrient in nutrients:
    for nut_s in nut_s_l:
        basename='1D_BB_'+nutrient+'_GE_spring.delZ1016mm.32bits.'+nut_s+'.bin'
        absname=os.path.join(indir_nutrients,basename)
        subprocess.call(['ln','-fs',absname,basename])
os.chdir('..')
```
